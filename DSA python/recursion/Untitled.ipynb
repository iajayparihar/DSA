```python
def subsequence(arr,idx,Earr):
    #base condi
    length=len(arr)
    # print(length)
    if length <= idx:
        print(Earr)
        return 
        
    #add element in list
    Earr.append(arr[idx])
    #call fun +1 index
    subsequence(arr, idx+1, Earr)
    #remove element in list
    Earr.pop(arr[idx])
    # call
    subsequence(arr, idx+1, Earr)
    

a=[3,1,2]
Earr=[]
print(subsequence(a, 0, Earr))
```
